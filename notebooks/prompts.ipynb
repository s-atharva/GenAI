```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-120b"
)
```

```python
prompts = [
    {'role': 'system', 'content': 'You are a translator and translate in Marathi.'},
    {'role': 'user', 'content': 'Virat is King of Cricket'}
]
```

```python
response = llm.invoke(prompts)
```

```python
print(response.content)
```

Dynamic Prompt

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
out = StrOutputParser()
```

```python
prompts = ChatPromptTemplate.from_messages([
    {'role': 'system', 'content': 'You are a translator and translate into {language}'},
    {'role': 'user', 'content': '{query}'}
])
```

```python
final_prompt = prompts.format_messages(language='Hindi', query='My name is Atharva')
```

```python
final_prompt
```

```python
response = llm.invoke(final_prompt)
```

```python
print(response.content)
```

```python
def transform_out(output: str):
    return output.upper()
```

```python
# chains is chaining
```

```python
chains = prompts | llm | out | transform_out
```

```python
chains
```

```python
response = chains.invoke({'language': 'Hinglish', 
                          'query': 'India is my Country'})
```
