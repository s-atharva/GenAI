```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(
    model='openai/gpt-oss-120b'
)
```

```python
from langchain.agents import create_agent
from langchain_core.tools import tool
from langgraph.checkpoint.memory import InMemorySaver
```

```python
@tool
def get_name_with_city(name, city):
    """
    This function returns employee's name with his city.
    args:
    name : name of employee
    city : name of city
    """
    return f"Your name is {name} and your city is {city}"
```

```python
@tool
def get_name_with_state(name, state):
    """
    This function returns employee's name with his state.
    args:
    name : name of employee
    state : name of state
    """
    return f"Your name is {name} and your city is {state}"
```

```python
agent = create_agent(
    model=llm,
    tools=[get_name_with_city, get_name_with_state],
    system_prompt="You are a manager who wants to show employees information based on requirements.",
    checkpointer=InMemorySaver()
)
```

```python
config = {'configurable': {'thread_id': 'm1'}}
```

```python
response = agent.invoke(
    {'messages': {'role': 'user', 'content': 'My name is Atharva and from Pune.'}},
    config=config
)
```

```python
print(response['messages'][-1].content)
```

```python
response = agent.invoke({'messages': [{'role': 'user', 'content': 'Tell me atharva city?'}]},
                        config=config)
```

```python
print(response['messages'][-1].content)
```
