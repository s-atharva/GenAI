```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(
    model='openai/gpt-oss-120b'
)
```

```python
text = "Hello, My name is Atharva Sarbhukan, Email is atharvasarbhukan@gmail.com and age is 23"
```

```python
response = llm.invoke(f'Please give me name, email and age from this text: {text}')
```

```python
print(response.content)
```

```python
from pydantic import BaseModel, Field
from typing import List
```

```python
class ResponseStructure(BaseModel):
    name: str = Field(description="Complete Name")
    email: str = Field(description="Student Email")
    age: int = Field(description="Student Age")
```

```python
structured_llm = llm.with_structured_output(ResponseStructure)
```

```python
response = structured_llm.invoke(f'Please give me name, email and age from this text: {text}')
```

```python
print(response)
```
