```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(
    model = 'openai/gpt-oss-120b'
)
```

```python
response = llm.invoke('what is the Capital of Delhi.')
print(response.content)
```

```python
response = llm.invoke('My name is Atharva')
print(response.content)
```

```python
history = []
while True:
    query = input('User: ')
    if query.lower() in ['quit', 'exit', '0']:
        print('GoodBye')
        break
    
    print(f'User: {query}')
    history.append({'role': 'user', 'content': query})
    
    response = llm.invoke(history)
    
    history.append({'role': 'ai', 'content': response.content})
    
    print(f'AI: {response.content}')
```
